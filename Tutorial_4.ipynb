```python
#import libraries
#import numpy as np
#import pandas as pd
import matplotlib.pyplot as plt
#from sklearn.model_selection import train_test_split
#from sklearn.preprocessing import StandardScaler #for scaling
#from sklearn.metrics import accuracy_score, classification_report
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img
import os
```

```python
save_folder = r"D:\PhD Nust 2026\First Semester Courses\Deep Learning\Tutorial 4"
os.makedirs(save_folder, exist_ok =True)
datagen = ImageDataGenerator(
    rotation_range = 40,
    shear_range = 0.2,
    zoom_range = 0.2,
    horizontal_flip = True,
    brightness_range = (0.5, 1.5)
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense

#step 2: Building neural netwrok model
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation = "relu"),
    Dense(64, activation = "relu"),
    Dense(10, activation = "softmax")
])
model.summary()
```

```python
img = load_img("/content/raccoon.jpg")
x = img_to_array(img)
x = x.reshape((1,) + x.shape)
```

```python
i = 0
for batch in datagen.flow(x, batch_size = 1, save_to_dir = save_folder, save_prefix = "raccoon", save_format = "jpeg"):

 augmented_image = batch[0].astype('uint8')
 plt.figure()
 plt.imshow(augmented_image)
 plt.axis('off')
 plt.show()

 i += 1
 if i > 40:
    break
```

```python
print(f"Augmented images are saved in thi folder: {os.path.abspath(save_folder)}")
```
